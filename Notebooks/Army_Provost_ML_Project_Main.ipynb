```python
# ==========================================================
# STEP 6 : BASIC DATASET OVERVIEW
# ==========================================================

def dataset_overview(df, dataset_name):
    """
    Display basic information about a dataset.
    """

    print("=" * 80)
    print(f"{dataset_name.upper()} DATASET OVERVIEW")
    print("=" * 80)

    # Shape
    print(f"\nShape:")
    print(f"Rows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")

    # Memory Usage
    memory_mb = df.memory_usage(deep=True).sum() / (1024 ** 2)

    print(f"\nMemory Usage:")
    print(f"{memory_mb:.2f} MB")

    # Column Information
    print("\nColumns and Data Types:\n")

    info_df = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str)
    })

    print(info_df.to_string(index=False))

    print("\n")


# ==========================================================
# CHICAGO DATASET
# ==========================================================

dataset_overview(
    chicago_df,
    "Chicago Crimes"
)

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

dataset_overview(
    montgomery_df,
    "Montgomery Police Dispatch"
)
```

```python
# ==========================================================
# STEP 7 : VIEW SAMPLE RECORDS
# ==========================================================

def display_samples(df, dataset_name, n=5):
    """
    Display the first and last few rows of a dataset.
    """

    print("=" * 80)
    print(f"{dataset_name.upper()} - FIRST {n} ROWS")
    print("=" * 80)

    display(df.head(n))

    print("\n")

    print("=" * 80)
    print(f"{dataset_name.upper()} - LAST {n} ROWS")
    print("=" * 80)

    display(df.tail(n))


# ==========================================================
# CHICAGO DATASET
# ==========================================================

display_samples(chicago_df, "Chicago Crimes")

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

display_samples(montgomery_df, "Montgomery Police Dispatch")
```

```python
# ==========================================================
# STEP 8 : MISSING VALUE ANALYSIS
# ==========================================================

def missing_value_summary(df, dataset_name):
    """
    Display missing value statistics for a dataset.
    """

    print("=" * 90)
    print(f"{dataset_name.upper()} - MISSING VALUE ANALYSIS")
    print("=" * 90)

    missing = df.isnull().sum()

    missing = missing[missing > 0]

    if len(missing) == 0:
        print("No missing values found.\n")
        return

    summary = pd.DataFrame({
        "Missing Values": missing,
        "Percentage (%)": (missing / len(df)) * 100
    })

    summary = summary.sort_values(
        by="Missing Values",
        ascending=False
    )

    print(summary)


# ==========================================================
# CHICAGO DATASET
# ==========================================================

missing_value_summary(
    chicago_df,
    "Chicago Crimes"
)

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

missing_value_summary(
    montgomery_df,
    "Montgomery Police Dispatch"
)
```

```python
# ==========================================================
# STEP 9 : DUPLICATE RECORD ANALYSIS
# ==========================================================

def duplicate_summary(df, dataset_name):
    """
    Display duplicate record statistics for a dataset.
    """

    print("=" * 80)
    print(f"{dataset_name.upper()} - DUPLICATE RECORD ANALYSIS")
    print("=" * 80)

    duplicate_count = df.duplicated().sum()
    duplicate_percentage = (duplicate_count / len(df)) * 100

    print(f"Total Rows            : {len(df):,}")
    print(f"Duplicate Rows        : {duplicate_count:,}")
    print(f"Duplicate Percentage  : {duplicate_percentage:.6f}%")
    print()


# ==========================================================
# CHICAGO DATASET
# ==========================================================

duplicate_summary(
    chicago_df,
    "Chicago Crimes"
)

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

duplicate_summary(
    montgomery_df,
    "Montgomery Police Dispatch"
)
```

```python
# ==========================================================
# STEP 10 : NUMERICAL DESCRIPTIVE STATISTICS
# ==========================================================

def numerical_summary(df, dataset_name):
    """
    Display descriptive statistics for numerical columns.
    """

    print("=" * 90)
    print(f"{dataset_name.upper()} - NUMERICAL SUMMARY")
    print("=" * 90)

    numeric_df = df.select_dtypes(include=["number"])

    if numeric_df.empty:
        print("No numerical columns found.\n")
        return

    display(numeric_df.describe().T)


# ==========================================================
# CHICAGO DATASET
# ==========================================================

numerical_summary(
    chicago_df,
    "Chicago Crimes"
)

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

numerical_summary(
    montgomery_df,
    "Montgomery Police Dispatch"
)
```

```python
# ==========================================================
# STEP 11 : CATEGORICAL COLUMN ANALYSIS
# ==========================================================

def categorical_summary(df, dataset_name, max_examples=10):
    """
    Display information about categorical columns.
    """

    print("=" * 100)
    print(f"{dataset_name.upper()} - CATEGORICAL COLUMN ANALYSIS")
    print("=" * 100)

    categorical_cols = df.select_dtypes(include=["object", "category"]).columns

    for col in categorical_cols:
        print(f"\nColumn: {col}")
        print(f"Unique Values: {df[col].nunique(dropna=True):,}")

        examples = df[col].dropna().unique()[:max_examples]

        print("Example Values:")
        for value in examples:
            print(f"  - {value}")

        print("-" * 60)


# ==========================================================
# CHICAGO DATASET
# ==========================================================

categorical_summary(chicago_df, "Chicago Crimes")

# ==========================================================
# MONTGOMERY DATASET
# ==========================================================

categorical_summary(montgomery_df, "Montgomery Police Dispatch")
```

```python
# ==========================================================
# STEP 12 : DATA CLEANING PIPELINE SKELETON
# ==========================================================

def clean_missing_values(df):
    """
    Handle missing values.
    (Implementation will be added in the next steps.)
    """
    return df


def convert_data_types(df):
    """
    Convert columns to appropriate data types.
    """
    return df


def standardize_text(df):
    """
    Standardize text formatting.
    """
    return df


def clean_coordinates(df):
    """
    Handle invalid coordinate values.
    """
    return df


def optimize_memory(df):
    """
    Reduce memory usage by optimizing data types.
    """
    return df


def data_cleaning_pipeline(df):
    """
    Execute all cleaning steps in sequence.
    """

    print("Starting data cleaning pipeline...")

    df = clean_missing_values(df)
    df = convert_data_types(df)
    df = standardize_text(df)
    df = clean_coordinates(df)
    df = optimize_memory(df)

    print("Data cleaning pipeline completed.")

    return df


print("=" * 70)
print("DATA CLEANING PIPELINE CREATED SUCCESSFULLY")
print("=" * 70)
```

```python
# ==========================================================
# STEP 12 : DATA QUALITY REPORT
# ==========================================================

import pandas as pd
import os

def generate_data_quality_report(df, dataset_name, save_path):
    """
    Generates a comprehensive data quality report.
    """

    report = pd.DataFrame({
        "Column": df.columns,
        "Data Type": df.dtypes.astype(str).values,
        "Missing Values": df.isnull().sum().values,
        "Missing (%)": ((df.isnull().sum() / len(df)) * 100).round(2).values,
        "Unique Values": df.nunique(dropna=True).values
    })

    report["Memory Usage (MB)"] = [
        round(df[col].memory_usage(deep=True) / (1024 ** 2), 2)
        for col in df.columns
    ]

    print("=" * 80)
    print(f"{dataset_name} DATA QUALITY REPORT")
    print("=" * 80)

    display(report)

    # Save report as CSV
    filename = os.path.join(
        save_path,
        f"{dataset_name.lower().replace(' ', '_')}_data_quality_report.csv"
    )

    report.to_csv(filename, index=False)

    print("\nReport saved to:")
    print(filename)

    return report


# ----------------------------------------------------------
# Generate reports
# ----------------------------------------------------------

chicago_report = generate_data_quality_report(
    chicago_df,
    "Chicago Crimes",
    REPORTS_PATH
)

montgomery_report = generate_data_quality_report(
    montgomery_df,
    "Montgomery Dispatch",
    REPORTS_PATH
)
```

```python
# ==========================================================
# SECTION 5 : DATA CLEANING
# STEP 14 : CLEAN INVALID COORDINATES
# ==========================================================

import numpy as np

print("=" * 80)
print("CLEANING INVALID COORDINATES")
print("=" * 80)

# ----------------------------------------------------------
# Count invalid coordinates BEFORE cleaning
# ----------------------------------------------------------

invalid_lat_before = (montgomery_df["Latitude"] == 0).sum()
invalid_lon_before = (montgomery_df["Longitude"] == 0).sum()

print("\nMontgomery Dataset")
print(f"Invalid Latitude values  : {invalid_lat_before:,}")
print(f"Invalid Longitude values : {invalid_lon_before:,}")

# ----------------------------------------------------------
# Replace invalid coordinates with NaN
# ----------------------------------------------------------

montgomery_df["Latitude"] = montgomery_df["Latitude"].replace(0, np.nan)
montgomery_df["Longitude"] = montgomery_df["Longitude"].replace(0, np.nan)

# ----------------------------------------------------------
# Count invalid coordinates AFTER cleaning
# ----------------------------------------------------------

invalid_lat_after = (montgomery_df["Latitude"] == 0).sum()
invalid_lon_after = (montgomery_df["Longitude"] == 0).sum()

print("\nAfter Cleaning")

print(f"Invalid Latitude values  : {invalid_lat_after:,}")
print(f"Invalid Longitude values : {invalid_lon_after:,}")

print("\nCoordinate cleaning completed successfully.")
```
