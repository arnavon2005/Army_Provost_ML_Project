```python
# ======================================================================
# STEP 6 — RESPONSE TEAM CAPABILITY COVERAGE VALIDATION
# ======================================================================

# ----------------------------------------------------------------------
# Collect all capabilities possessed by each team
# ----------------------------------------------------------------------

def extract_team_capabilities(row):
    """
    Combine primary and secondary capabilities for one response team.
    """

    capabilities = {row["Primary_Capability"]}

    secondary = [
        cap.strip()
        for cap in str(row["Secondary_Capabilities"]).split(";")
        if cap.strip()
    ]

    capabilities.update(secondary)

    return capabilities


response_team_df["All_Capabilities"] = response_team_df.apply(
    extract_team_capabilities,
    axis=1
)

# ----------------------------------------------------------------------
# Required PRIMARY capabilities coming from the existing DSS mapping
# ----------------------------------------------------------------------

required_primary_capabilities = set(
    response_capability_df["Primary_Required_Capability"]
)

# Capabilities represented somewhere in the team registry
registry_capabilities = set()

for capabilities in response_team_df["All_Capabilities"]:
    registry_capabilities.update(capabilities)

missing_required_capabilities = (
    required_primary_capabilities - registry_capabilities
)

# ----------------------------------------------------------------------
# Coverage Report
# ----------------------------------------------------------------------

coverage_records = []

for capability in sorted(required_primary_capabilities):

    matching_teams = response_team_df[
        response_team_df["All_Capabilities"].apply(
            lambda x: capability in x
        )
    ]

    available_matching_teams = matching_teams[
        (matching_teams["Status"] == "AVAILABLE")
        &
        (matching_teams["Readiness_Level"] != "UNAVAILABLE")
    ]

    coverage_records.append(
        {
            "Required_Capability": capability,
            "Total_Capable_Teams": len(matching_teams),
            "Currently_Deployable_Teams": len(available_matching_teams),
            "Capable_Team_IDs": ", ".join(
                matching_teams["Team_ID"].tolist()
            ),
            "Deployable_Team_IDs": ", ".join(
                available_matching_teams["Team_ID"].tolist()
            )
        }
    )

capability_coverage_df = pd.DataFrame(coverage_records)

print("=" * 110)
print("RESPONSE TEAM CAPABILITY COVERAGE VALIDATION")
print("=" * 110)

display(capability_coverage_df)

print("\nCoverage Summary")
print("-" * 60)

print(
    "DSS Primary Capabilities Required :",
    len(required_primary_capabilities)
)

print(
    "Capabilities Present in Registry  :",
    len(required_primary_capabilities - missing_required_capabilities)
)

print(
    "Missing Required Capabilities     :",
    len(missing_required_capabilities)
)

if len(missing_required_capabilities) == 0:
    print("\n✓ Every primary DSS-required capability is represented in the registry.")
else:
    print("\n✗ Missing capabilities:")
    for capability in sorted(missing_required_capabilities):
        print("  -", capability)

# ----------------------------------------------------------------------
# Current Deployability Check
# ----------------------------------------------------------------------

no_deployable_team = capability_coverage_df[
    capability_coverage_df["Currently_Deployable_Teams"] == 0
]

if len(no_deployable_team) == 0:
    print(
        "✓ Every primary capability currently has at least one "
        "deployable response team."
    )
else:
    print(
        "⚠ Some capabilities currently have no immediately "
        "deployable response team:"
    )

    display(
        no_deployable_team[
            [
                "Required_Capability",
                "Capable_Team_IDs"
            ]
        ]
    )

print("\n" + "=" * 110)
print("CAPABILITY COVERAGE VALIDATION COMPLETE")
print("=" * 110)
```

```python
# ======================================================================
# STEP 10 — ELIGIBLE RESPONSE TEAM FILTER
# ======================================================================

def get_team_capabilities(row):
    """
    Return the complete capability set for a response team.
    """

    capabilities = {row["Primary_Capability"]}

    secondary = [
        cap.strip()
        for cap in str(row["Secondary_Capabilities"]).split(";")
        if cap.strip()
    ]

    capabilities.update(secondary)

    return capabilities


def filter_eligible_response_teams(
    team_registry,
    required_capability
):
    """
    Filter teams that are immediately eligible for deployment.

    Eligibility Rules:
    1. Status must be AVAILABLE
    2. Readiness must not be UNAVAILABLE
    3. Team must possess the required capability
       either as a primary or secondary capability
    """

    eligible_records = []

    for _, team in team_registry.iterrows():

        team_capabilities = get_team_capabilities(team)

        is_available = (
            team["Status"] == "AVAILABLE"
        )

        is_ready_enough = (
            team["Readiness_Level"] != "UNAVAILABLE"
        )

        has_required_capability = (
            required_capability in team_capabilities
        )

        if (
            is_available
            and is_ready_enough
            and has_required_capability
        ):
            eligible_records.append(team)

    if eligible_records:
        eligible_df = pd.DataFrame(eligible_records).reset_index(drop=True)
    else:
        eligible_df = pd.DataFrame(columns=team_registry.columns)

    return eligible_df


# ----------------------------------------------------------------------
# TEST CASE — PERSONNEL SAFETY
# ----------------------------------------------------------------------

test_required_capability = "Personnel Safety"

eligible_test_df = filter_eligible_response_teams(
    response_team_df,
    test_required_capability
)

print("=" * 100)
print("ELIGIBLE RESPONSE TEAM FILTER — TEST")
print("=" * 100)

print("\nRequired Capability:")
print(test_required_capability)

print("\nEligible Teams:")
display(
    eligible_test_df[
        [
            "Team_ID",
            "Team_Name",
            "Status",
            "Current_Zone",
            "Primary_Capability",
            "Secondary_Capabilities",
            "Readiness_Level",
            "Personnel_Strength",
            "Vehicle_Available"
        ]
    ]
)

print("\nEligible Team Count:", len(eligible_test_df))

print("\nExpected Logic:")
print("- RT-01 should qualify through SECONDARY Personnel Safety capability.")
print("- RT-02 should qualify through PRIMARY Personnel Safety capability.")
print("- RT-05 should qualify through PRIMARY Personnel Safety capability.")
print("- ASSIGNED / OFFLINE teams must not appear.")

print("\n" + "=" * 100)
print("ELIGIBILITY FILTER TEST COMPLETE")
print("=" * 100)
```

```python
# ======================================================================
# STEP 11 — TEAM SUITABILITY SCORING FUNCTION
# ======================================================================

def personnel_strength_score(strength):
    """
    Convert personnel strength into a small supporting score.
    """
    if strength >= 7:
        return 5
    elif strength == 6:
        return 4
    elif strength == 5:
        return 3
    elif strength == 4:
        return 2
    else:
        return 1


def score_response_team(
    team,
    required_capability,
    incident_zone
):
    """
    Calculate suitability score for an already eligible response team.

    Maximum score = 100
    """

    score = 0
    score_breakdown = {}

    # --------------------------------------------------------------
    # 1. Capability Match
    # --------------------------------------------------------------

    if team["Primary_Capability"] == required_capability:
        capability_score = 40
        capability_match_type = "PRIMARY"

    else:
        capability_score = 30
        capability_match_type = "SECONDARY"

    score += capability_score
    score_breakdown["Capability_Score"] = capability_score

    # --------------------------------------------------------------
    # 2. Readiness
    # --------------------------------------------------------------

    if team["Readiness_Level"] == "READY":
        readiness_score = 25

    elif team["Readiness_Level"] == "LIMITED":
        readiness_score = 12

    else:
        readiness_score = 0

    score += readiness_score
    score_breakdown["Readiness_Score"] = readiness_score

    # --------------------------------------------------------------
    # 3. Zone Relationship
    # --------------------------------------------------------------

    if team["Current_Zone"] == incident_zone:
        zone_score = 20
        zone_match = "SAME ZONE"

    else:
        zone_score = 10
        zone_match = "DIFFERENT ZONE"

    score += zone_score
    score_breakdown["Zone_Score"] = zone_score

    # --------------------------------------------------------------
    # 4. Vehicle Availability
    # --------------------------------------------------------------

    vehicle_score = 10 if team["Vehicle_Available"] else 0

    score += vehicle_score
    score_breakdown["Vehicle_Score"] = vehicle_score

    # --------------------------------------------------------------
    # 5. Personnel Strength
    # --------------------------------------------------------------

    strength_score = personnel_strength_score(
        team["Personnel_Strength"]
    )

    score += strength_score
    score_breakdown["Personnel_Score"] = strength_score

    # --------------------------------------------------------------
    # Return result
    # --------------------------------------------------------------

    return {
        "Team_ID": team["Team_ID"],
        "Team_Name": team["Team_Name"],
        "Current_Zone": team["Current_Zone"],
        "Capability_Match": capability_match_type,
        "Readiness_Level": team["Readiness_Level"],
        "Zone_Relationship": zone_match,
        "Vehicle_Available": team["Vehicle_Available"],
        "Personnel_Strength": team["Personnel_Strength"],
        **score_breakdown,
        "Suitability_Score": score
    }


def rank_eligible_response_teams(
    eligible_teams,
    required_capability,
    incident_zone
):
    """
    Score and rank all eligible response teams.
    """

    ranking_records = []

    for _, team in eligible_teams.iterrows():

        result = score_response_team(
            team,
            required_capability,
            incident_zone
        )

        ranking_records.append(result)

    ranking_df = pd.DataFrame(ranking_records)

    if not ranking_df.empty:

        ranking_df = ranking_df.sort_values(
            by="Suitability_Score",
            ascending=False
        ).reset_index(drop=True)

        ranking_df.insert(
            0,
            "Rank",
            range(1, len(ranking_df) + 1)
        )

    return ranking_df


# ======================================================================
# TEST CASE
# ======================================================================

test_required_capability = "Personnel Safety"
test_incident_zone = "Zone B"

eligible_test_df = filter_eligible_response_teams(
    response_team_df,
    test_required_capability
)

ranking_test_df = rank_eligible_response_teams(
    eligible_test_df,
    test_required_capability,
    test_incident_zone
)

print("=" * 110)
print("RESPONSE TEAM SUITABILITY RANKING — TEST")
print("=" * 110)

print("\nRequired Capability :", test_required_capability)
print("Incident Zone       :", test_incident_zone)

print("\nRanked Teams:")

display(
    ranking_test_df[
        [
            "Rank",
            "Team_ID",
            "Capability_Match",
            "Readiness_Level",
            "Zone_Relationship",
            "Vehicle_Available",
            "Personnel_Strength",
            "Capability_Score",
            "Readiness_Score",
            "Zone_Score",
            "Vehicle_Score",
            "Personnel_Score",
            "Suitability_Score"
        ]
    ]
)

if not ranking_test_df.empty:

    print("\nRecommended Primary Team:")
    print(
        ranking_test_df.iloc[0]["Team_ID"],
        "- Score:",
        ranking_test_df.iloc[0]["Suitability_Score"]
    )

    if len(ranking_test_df) > 1:
        print("\nRecommended Alternate Team:")
        print(
            ranking_test_df.iloc[1]["Team_ID"],
            "- Score:",
            ranking_test_df.iloc[1]["Suitability_Score"]
        )

print("\nExpected Result:")
print("RT-02 should rank FIRST because:")
print("- Personnel Safety is its PRIMARY capability")
print("- It is READY")
print("- It is already in Zone B")
print("- A vehicle is available")
print("- It has strong personnel strength")

print("\n" + "=" * 110)
print("TEAM SUITABILITY RANKING TEST COMPLETE")
print("=" * 110)
```

```python
# ======================================================================
# STEP 12 — DSS RESPONSE TO RESOURCE ALLOCATION BRIDGE
# ======================================================================

def get_required_capabilities(operational_response):
    """
    Convert an existing DSS operational-response category into
    its required primary and secondary capabilities.
    """

    match = response_capability_df[
        response_capability_df["Operational_Response"]
        == operational_response
    ]

    if match.empty:
        raise ValueError(
            f"No capability mapping found for operational response: "
            f"{operational_response}"
        )

    row = match.iloc[0]

    primary_capability = row[
        "Primary_Required_Capability"
    ]

    secondary_capabilities = [
        cap.strip()
        for cap in row[
            "Secondary_Required_Capabilities"
        ].split(";")
        if cap.strip()
    ]

    return {
        "Primary_Required_Capability": primary_capability,
        "Secondary_Required_Capabilities": secondary_capabilities
    }


def allocate_response_resource(
    operational_response,
    incident_zone,
    team_registry=response_team_df
):
    """
    Perform resource recommendation using an operational-response
    category produced by the existing DSS.

    Flow:
    DSS Operational Response
        ↓
    Required Capability
        ↓
    Eligibility Filtering
        ↓
    Suitability Ranking
        ↓
    Primary + Alternate Recommendation
    """

    capability_requirements = get_required_capabilities(
        operational_response
    )

    required_capability = capability_requirements[
        "Primary_Required_Capability"
    ]

    eligible_teams = filter_eligible_response_teams(
        team_registry,
        required_capability
    )

    ranked_teams = rank_eligible_response_teams(
        eligible_teams,
        required_capability,
        incident_zone
    )

    primary_team = None
    alternate_team = None

    if len(ranked_teams) >= 1:
        primary_team = ranked_teams.iloc[0].to_dict()

    if len(ranked_teams) >= 2:
        alternate_team = ranked_teams.iloc[1].to_dict()

    return {
        "Operational_Response": operational_response,
        "Required_Primary_Capability": required_capability,
        "Required_Secondary_Capabilities":
            capability_requirements[
                "Secondary_Required_Capabilities"
            ],
        "Incident_Zone": incident_zone,
        "Eligible_Team_Count": len(eligible_teams),
        "Primary_Recommendation": primary_team,
        "Alternate_Recommendation": alternate_team,
        "Full_Ranking": ranked_teams
    }


# ======================================================================
# TEST — USING AN EXISTING DSS RESPONSE CATEGORY
# ======================================================================

test_operational_response = (
    "Personnel Safety / Provost Response"
)

test_incident_zone = "Zone B"

allocation_test = allocate_response_resource(
    operational_response=test_operational_response,
    incident_zone=test_incident_zone
)

print("=" * 110)
print("DSS → RESOURCE ALLOCATION BRIDGE TEST")
print("=" * 110)

print("\nExisting DSS Operational Response:")
print(allocation_test["Operational_Response"])

print("\nDerived Primary Capability:")
print(allocation_test["Required_Primary_Capability"])

print("\nDerived Secondary Capabilities:")
for capability in allocation_test[
    "Required_Secondary_Capabilities"
]:
    print(" -", capability)

print("\nIncident Zone:")
print(allocation_test["Incident_Zone"])

print("\nEligible Team Count:")
print(allocation_test["Eligible_Team_Count"])


if allocation_test["Primary_Recommendation"]:

    primary = allocation_test[
        "Primary_Recommendation"
    ]

    print("\nPRIMARY RECOMMENDATION")
    print("-" * 50)
    print("Team ID           :", primary["Team_ID"])
    print("Team Name         :", primary["Team_Name"])
    print("Current Zone      :", primary["Current_Zone"])
    print("Capability Match  :", primary["Capability_Match"])
    print("Readiness         :", primary["Readiness_Level"])
    print(
        "Suitability Score :",
        f'{primary["Suitability_Score"]}/100'
    )


if allocation_test["Alternate_Recommendation"]:

    alternate = allocation_test[
        "Alternate_Recommendation"
    ]

    print("\nALTERNATE RECOMMENDATION")
    print("-" * 50)
    print("Team ID           :", alternate["Team_ID"])
    print("Team Name         :", alternate["Team_Name"])
    print("Current Zone      :", alternate["Current_Zone"])
    print("Capability Match  :", alternate["Capability_Match"])
    print("Readiness         :", alternate["Readiness_Level"])
    print(
        "Suitability Score :",
        f'{alternate["Suitability_Score"]}/100'
    )


print("\nFull Team Ranking:")
display(
    allocation_test["Full_Ranking"][
        [
            "Rank",
            "Team_ID",
            "Capability_Match",
            "Readiness_Level",
            "Zone_Relationship",
            "Suitability_Score"
        ]
    ]
)

print("\nIMPORTANT:")
print("This test starts from an operational-response category")
print("already produced by the existing DSS.")
print("The resource layer does NOT replace the existing DSS.")

print("\n" + "=" * 110)
print("DSS TO RESOURCE ALLOCATION BRIDGE TEST COMPLETE")
print("=" * 110)
```

```python
# ======================================================================
# STEP 15 — RESPONSE TEAM ASSIGNMENT STATE MANAGEMENT
# ======================================================================

from datetime import datetime


def assign_response_team(
    team_registry,
    team_id,
    incident_id
):
    """
    Assign an AVAILABLE response team to a simulated incident.

    Returns a NEW registry copy.
    The original DataFrame is not modified in-place.
    """

    updated_registry = team_registry.copy(deep=True)

    team_match = updated_registry[
        updated_registry["Team_ID"] == team_id
    ]

    if team_match.empty:
        raise ValueError(
            f"Team ID '{team_id}' does not exist in the registry."
        )

    team_index = team_match.index[0]

    current_status = updated_registry.loc[
        team_index,
        "Status"
    ]

    readiness = updated_registry.loc[
        team_index,
        "Readiness_Level"
    ]

    if current_status != "AVAILABLE":
        raise ValueError(
            f"{team_id} cannot be assigned because its "
            f"current status is {current_status}."
        )

    if readiness == "UNAVAILABLE":
        raise ValueError(
            f"{team_id} cannot be assigned because its "
            "readiness level is UNAVAILABLE."
        )

    updated_registry.loc[
        team_index,
        "Status"
    ] = "ASSIGNED"

    updated_registry.loc[
        team_index,
        "Current_Assignment"
    ] = incident_id

    updated_registry.loc[
        team_index,
        "Last_Updated"
    ] = datetime.now()

    return updated_registry


def release_response_team(
    team_registry,
    team_id
):
    """
    Release a previously ASSIGNED response team.

    Returns a NEW registry copy.
    """

    updated_registry = team_registry.copy(deep=True)

    team_match = updated_registry[
        updated_registry["Team_ID"] == team_id
    ]

    if team_match.empty:
        raise ValueError(
            f"Team ID '{team_id}' does not exist in the registry."
        )

    team_index = team_match.index[0]

    current_status = updated_registry.loc[
        team_index,
        "Status"
    ]

    if current_status != "ASSIGNED":
        raise ValueError(
            f"{team_id} cannot be released because its "
            f"current status is {current_status}."
        )

    updated_registry.loc[
        team_index,
        "Status"
    ] = "AVAILABLE"

    updated_registry.loc[
        team_index,
        "Current_Assignment"
    ] = "None"

    updated_registry.loc[
        team_index,
        "Last_Updated"
    ] = datetime.now()

    return updated_registry


# ======================================================================
# TEST — ASSIGN AND RELEASE RT-02
# ======================================================================

state_test_registry = response_team_df.copy(deep=True)

print("=" * 110)
print("RESPONSE TEAM ASSIGNMENT STATE MANAGEMENT TEST")
print("=" * 110)

print("\nINITIAL STATE")

initial_rt02 = state_test_registry[
    state_test_registry["Team_ID"] == "RT-02"
].iloc[0]

print("Team ID            :", initial_rt02["Team_ID"])
print("Status             :", initial_rt02["Status"])
print(
    "Current Assignment :",
    initial_rt02["Current_Assignment"]
)

# ----------------------------------------------------------------------
# Assign team
# ----------------------------------------------------------------------

state_test_registry = assign_response_team(
    team_registry=state_test_registry,
    team_id="RT-02",
    incident_id="SIM-INC-0100"
)

assigned_rt02 = state_test_registry[
    state_test_registry["Team_ID"] == "RT-02"
].iloc[0]

print("\nAFTER ASSIGNMENT")
print("Team ID            :", assigned_rt02["Team_ID"])
print("Status             :", assigned_rt02["Status"])
print(
    "Current Assignment :",
    assigned_rt02["Current_Assignment"]
)

# ----------------------------------------------------------------------
# Confirm assigned team is excluded from a new allocation
# ----------------------------------------------------------------------

post_assignment_allocation = allocate_response_resource(
    operational_response="Personnel Safety / Provost Response",
    incident_zone="Zone B",
    team_registry=state_test_registry
)

next_team = (
    post_assignment_allocation[
        "Primary_Recommendation"
    ]["Team_ID"]
    if post_assignment_allocation[
        "Primary_Recommendation"
    ]
    else None
)

print("\nNEXT ALLOCATION AFTER RT-02 ASSIGNMENT")
print("Recommended Team :", next_team)

# ----------------------------------------------------------------------
# Release team
# ----------------------------------------------------------------------

state_test_registry = release_response_team(
    team_registry=state_test_registry,
    team_id="RT-02"
)

released_rt02 = state_test_registry[
    state_test_registry["Team_ID"] == "RT-02"
].iloc[0]

print("\nAFTER RELEASE")
print("Team ID            :", released_rt02["Team_ID"])
print("Status             :", released_rt02["Status"])
print(
    "Current Assignment :",
    released_rt02["Current_Assignment"]
)

# ----------------------------------------------------------------------
# Validation
# ----------------------------------------------------------------------

assignment_valid = (
    assigned_rt02["Status"] == "ASSIGNED"
    and assigned_rt02["Current_Assignment"]
    == "SIM-INC-0100"
)

fallback_valid = (
    next_team is not None
    and next_team != "RT-02"
)

release_valid = (
    released_rt02["Status"] == "AVAILABLE"
    and released_rt02["Current_Assignment"] == "None"
)

print("\nValidation")
print("-" * 60)

print(
    "Assignment State Update :",
    "✓ PASS" if assignment_valid else "✗ FAIL"
)

print(
    "Assigned Team Exclusion :",
    "✓ PASS" if fallback_valid else "✗ FAIL"
)

print(
    "Team Release            :",
    "✓ PASS" if release_valid else "✗ FAIL"
)

if assignment_valid and fallback_valid and release_valid:
    print(
        "\n✓ RESPONSE TEAM STATE MANAGEMENT "
        "TEST PASSED."
    )
else:
    print(
        "\n✗ RESPONSE TEAM STATE MANAGEMENT "
        "TEST FAILED."
    )

print("\nIMPORTANT:")
print(
    "The baseline Response_Team_Registry.csv "
    "was NOT modified by this test."
)

print("\n" + "=" * 110)
print("RESPONSE TEAM ASSIGNMENT STATE MANAGEMENT TEST COMPLETE")
print("=" * 110)
```

```python
# ======================================================================
# STEP 16 — HUMAN OPERATOR CONFIRM / OVERRIDE LOGIC
# ======================================================================

from datetime import datetime
import uuid


def process_operator_decision(
    allocation_result,
    team_registry,
    operator_uid,
    action,
    selected_team_id=None,
    override_reason=None,
    incident_id=None
):
    """
    Process a human operator decision after resource recommendation.

    action:
        CONFIRM  -> accept DSS primary recommendation
        OVERRIDE -> select another currently eligible team

    Returns:
        decision_record
        updated_registry
    """

    action = str(action).strip().upper()

    if action not in ["CONFIRM", "OVERRIDE"]:
        raise ValueError(
            "Operator action must be CONFIRM or OVERRIDE."
        )

    recommended_team = allocation_result[
        "Primary_Recommendation"
    ]

    if recommended_team is None:
        raise ValueError(
            "No primary team recommendation exists. "
            "Operator confirmation/override cannot proceed."
        )

    recommended_team_id = recommended_team["Team_ID"]

    # ------------------------------------------------------------------
    # CONFIRM
    # ------------------------------------------------------------------

    if action == "CONFIRM":

        final_team_id = recommended_team_id
        decision_type = "CONFIRMED"
        override_reason_final = None

    # ------------------------------------------------------------------
    # OVERRIDE
    # ------------------------------------------------------------------

    else:

        if selected_team_id is None:
            raise ValueError(
                "selected_team_id is required for OVERRIDE."
            )

        if selected_team_id == recommended_team_id:
            raise ValueError(
                "Override team cannot be the same as "
                "the recommended team."
            )

        ranked_teams = allocation_result["Full_Ranking"]

        eligible_team_ids = set(
            ranked_teams["Team_ID"].tolist()
        )

        if selected_team_id not in eligible_team_ids:
            raise ValueError(
                f"{selected_team_id} is not currently an "
                "eligible deployment option."
            )

        if (
            override_reason is None
            or not str(override_reason).strip()
        ):
            raise ValueError(
                "An override reason is required."
            )

        final_team_id = selected_team_id
        decision_type = "OVERRIDDEN"
        override_reason_final = str(
            override_reason
        ).strip()

    # ------------------------------------------------------------------
    # Incident / Decision IDs
    # ------------------------------------------------------------------

    if incident_id is None:
        incident_id = (
            "SIM-INC-"
            + datetime.now().strftime("%Y%m%d%H%M%S")
        )

    decision_id = (
        "DEC-"
        + uuid.uuid4().hex[:10].upper()
    )

    decision_timestamp = datetime.now()

    # ------------------------------------------------------------------
    # Assign final human-selected team
    # ------------------------------------------------------------------

    updated_registry = assign_response_team(
        team_registry=team_registry,
        team_id=final_team_id,
        incident_id=incident_id
    )

    # ------------------------------------------------------------------
    # Find selected team's ranking information
    # ------------------------------------------------------------------

    selected_row = allocation_result[
        "Full_Ranking"
    ][
        allocation_result["Full_Ranking"]["Team_ID"]
        == final_team_id
    ].iloc[0]

    # ------------------------------------------------------------------
    # Decision record
    # ------------------------------------------------------------------

    decision_record = {
        "Decision_ID": decision_id,
        "Incident_ID": incident_id,
        "Timestamp": decision_timestamp,
        "Operator_UID": operator_uid,

        "Operational_Response":
            allocation_result["Operational_Response"],

        "Incident_Zone":
            allocation_result["Incident_Zone"],

        "Required_Capability":
            allocation_result[
                "Required_Primary_Capability"
            ],

        "Recommended_Team_ID":
            recommended_team_id,

        "Recommended_Team_Score":
            recommended_team[
                "Suitability_Score"
            ],

        "Operator_Action":
            decision_type,

        "Selected_Team_ID":
            final_team_id,

        "Selected_Team_Score":
            selected_row[
                "Suitability_Score"
            ],

        "Override_Reason":
            override_reason_final
    }

    return decision_record, updated_registry


# ======================================================================
# TEST 1 — OPERATOR CONFIRMS DSS RECOMMENDATION
# ======================================================================

confirm_test_registry = response_team_df.copy(deep=True)

confirm_allocation = allocate_response_resource(
    operational_response="Personnel Safety / Provost Response",
    incident_zone="Zone B",
    team_registry=confirm_test_registry
)

confirm_record, confirm_updated_registry = (
    process_operator_decision(
        allocation_result=confirm_allocation,
        team_registry=confirm_test_registry,
        operator_uid="TEST-OPERATOR-01",
        action="CONFIRM",
        incident_id="SIM-INC-CONFIRM-001"
    )
)

print("=" * 110)
print("HUMAN OPERATOR DECISION TEST")
print("=" * 110)

print("\nTEST 1 — CONFIRM RECOMMENDATION")
print("-" * 60)

print(
    "Recommended Team :",
    confirm_record["Recommended_Team_ID"]
)

print(
    "Selected Team    :",
    confirm_record["Selected_Team_ID"]
)

print(
    "Operator Action  :",
    confirm_record["Operator_Action"]
)

print(
    "Decision ID      :",
    confirm_record["Decision_ID"]
)

confirmed_team_status = confirm_updated_registry.loc[
    confirm_updated_registry["Team_ID"]
    == confirm_record["Selected_Team_ID"],
    "Status"
].iloc[0]

print(
    "Final Team Status:",
    confirmed_team_status
)


# ======================================================================
# TEST 2 — OPERATOR OVERRIDES DSS RECOMMENDATION
# ======================================================================

override_test_registry = response_team_df.copy(deep=True)

override_allocation = allocate_response_resource(
    operational_response="Personnel Safety / Provost Response",
    incident_zone="Zone B",
    team_registry=override_test_registry
)

override_record, override_updated_registry = (
    process_operator_decision(
        allocation_result=override_allocation,
        team_registry=override_test_registry,
        operator_uid="TEST-OPERATOR-01",
        action="OVERRIDE",
        selected_team_id="RT-01",
        override_reason=(
            "Operator selected alternate team for "
            "simulated operational coordination."
        ),
        incident_id="SIM-INC-OVERRIDE-001"
    )
)

print("\nTEST 2 — OVERRIDE RECOMMENDATION")
print("-" * 60)

print(
    "Recommended Team :",
    override_record["Recommended_Team_ID"]
)

print(
    "Selected Team    :",
    override_record["Selected_Team_ID"]
)

print(
    "Operator Action  :",
    override_record["Operator_Action"]
)

print(
    "Override Reason  :",
    override_record["Override_Reason"]
)

print(
    "Decision ID      :",
    override_record["Decision_ID"]
)

overridden_team_status = (
    override_updated_registry.loc[
        override_updated_registry["Team_ID"]
        == override_record["Selected_Team_ID"],
        "Status"
    ].iloc[0]
)

print(
    "Final Team Status:",
    overridden_team_status
)


# ======================================================================
# VALIDATION
# ======================================================================

confirm_pass = (
    confirm_record["Recommended_Team_ID"]
    == confirm_record["Selected_Team_ID"]
    and confirm_record["Operator_Action"]
    == "CONFIRMED"
    and confirmed_team_status == "ASSIGNED"
)

override_pass = (
    override_record["Recommended_Team_ID"]
    != override_record["Selected_Team_ID"]
    and override_record["Operator_Action"]
    == "OVERRIDDEN"
    and overridden_team_status == "ASSIGNED"
    and override_record["Override_Reason"] is not None
)

print("\nValidation")
print("-" * 60)

print(
    "Operator Confirmation :",
    "✓ PASS" if confirm_pass else "✗ FAIL"
)

print(
    "Operator Override     :",
    "✓ PASS" if override_pass else "✗ FAIL"
)

if confirm_pass and override_pass:
    print(
        "\n✓ HUMAN-IN-THE-LOOP DECISION LOGIC TEST PASSED."
    )
else:
    print(
        "\n✗ HUMAN-IN-THE-LOOP DECISION LOGIC TEST FAILED."
    )

print("\nIMPORTANT:")
print("The DSS only recommends the response resource.")
print("The human operator determines the final selected team.")
print("Override decisions require an explicit reason.")
print("The baseline Response_Team_Registry.csv remains unchanged.")

print("\n" + "=" * 110)
print("HUMAN OPERATOR CONFIRM / OVERRIDE TEST COMPLETE")
print("=" * 110)
```
